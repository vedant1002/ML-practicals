```python
# K means clustering using silhoutte method
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# Load the Iris dataset
iris = load_iris()
X = iris.data
# Scale the features so they are on comparable ranges
X_scaled = StandardScaler().fit_transform(X)
```

```python
# Try cluster counts from 2 to 10
cluster_counts = range(2, 11)
scores = []
```

```python
for k in cluster_counts:
    # Create and fit a K-Means model with k clusters
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    # Measure how well-separated the clusters are
    score = silhouette_score(X_scaled, labels)
    scores.append(score)
    # Display the score for this value of k
    print(f"k = {k}, silhouette score = {score:.3f}")
```

```python
# Find the cluster count with the highest silhouette score
best_k = cluster_counts[scores.index(max(scores))]
print(f"\nBest number of clusters: {best_k}")
```

```python
# Fit the final model using the selected cluster count
final_model = KMeans(n_clusters=best_k, random_state=42, n_init=10)
final_labels = final_model.fit_predict(X_scaled)
```

```python
# Show the cluster assigned to each data point
print("Cluster labels:")
print(final_labels)
```
