```python
# Install scikit-learn first if needed:
# pip install scikit-learn
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import numpy as np
```

```python
# Load the Iris flower dataset
iris = load_iris()
X = iris.data
```

```python
# Scale the measurements so each feature has a similar range
X_scaled = StandardScaler().fit_transform(X)
```

```python
# Create DBSCAN: nearby points within 0.8 are neighbors;
# a cluster needs at least 5 points in that neighborhood
model = DBSCAN(eps=0.8, min_samples=5)
```

```python
# Assign a cluster label to every flower
labels = model.fit_predict(X_scaled)
```

```python
# Find the distinct labels; -1 means noise
cluster_labels = sorted(set(labels))
print("Cluster labels:", cluster_labels)
```

```python
# Count points in each cluster and the noise group
for label in cluster_labels:
  count = np.sum(labels == label)
  if label == -1:
    print(f"Noise points: {count}")
else:
    print(f"Cluster {label}: {count} points")
```
