```python
import pandas as pd
import numpy as np
import seaborn as sns

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# 1.Load dataset
df = sns.load_dataset('titanic')

print(df.head())

print(df.info())
```

```python
# 2.Select Features
data = df[['survived','pclass','who','sex','age','sibsp','parch','fare','embarked']].copy()

print("\n Missing Values Before Processing: ")
print(data.isnull().sum())
```

```python
# 3.Handle missing values
# Numerical columns
num_cols = ['age','fare']

num_imputer = SimpleImputer(strategy='mean')
data[num_cols] = num_imputer.fit_transform(data[num_cols])

#Categorical column
cat_cols = ['embarked']

cat_imputer = SimpleImputer(strategy='most_frequent')
data[cat_cols] = cat_imputer.fit_transform(data[cat_cols])

print("\n Missing Values After Imputation")
print(data.isnull().sum())
```

```python
# 4.Encode Categorical Variables
le_sex = LabelEncoder()
le_embarked = LabelEncoder()

data['sex'] = le_sex.fit_transform(data['sex'])
data['embarked'] = le_embarked.fit_transform(data['embarked'])

print("\n Encoded Data")
print(data.head())
```

```python
# 5.Scale Numerical Features
scaler = StandardScaler()

scale_cols = ['age','fare']

data[scale_cols] = scaler.fit_transform(data[scale_cols])

print("\n Scaled Data")
print(data.head())
```

```python
# 6.Final Model
X = data.drop('survived', axis=1)
y = data['survived']

print("\n Feature Matrix Shape: ", X.shape)
print("\n Target Variable shape: ", y.shape)

print("\n Final Prepared Dataset: ")
print(X.head())
```

```python
#Optional: Save cleaned dataset
cleaned_data = pd.concat([X,y], axis=1)
cleaned_data.to_csv('titanic_cleaned_dataset.csv', index=False)

print("\n Cleaned dataset saved as 'titanic_cleaned_dataset.csv'")
```
