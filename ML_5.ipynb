```python
# Import libraries
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import dataset
from sklearn.datasets import load_breast_cancer
```

```python
# Import preprocessing and model selection tools
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Import evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
# Load the Breast Cancer dataset
data = load_breast_cancer()
```

```python
# Store input features in X
X = pd.DataFrame(data.data, columns=data.feature_names)

# Store target variable in y
y = pd.Series(data.target, name="target")
```

```python
# Display dataset information
print("Dataset shape:", X.shape)
print("\n First five rows:")
print(X.head())

# Check for missing values
print("\n Missing values:")
print(X.isnull().sum().sum())
```

```python
# CHECK MULTICOLLINEARITY

# Calculate correlation matrix
correlation_matrix = X.corr()

# Plot correlation heatmap
plt.figure(figsize=(14,10))
sns.heatmap(correlation_matrix, cmap='coolwarm', center=0)
plt.title("Correlation Matrix of features")
plt.show()
```

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
# RIDGE-LIKE LOGISTIC REGRESSION

# Create L2 regularized Logistic Regression pipeline
ridge_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(penalty="l2", C=1.0, max_iter=5000))
])
```

```python
# Train Ridge-like model
ridge_model.fit(X_train, y_train)

# Predict training data
ridge_train_pred = ridge_model.predict(X_train)

# Predict testing data
ridge_test_pred = ridge_model.predict(X_test)
```

```python
# Calculate training accuracy
ridge_train_accuracy = accuracy_score(y_train, ridge_train_pred)

# Calculate testing accuracy
ridge_test_accuracy = accuracy_score(y_test, ridge_test_pred)
```

```python
# Display Ridge results
print("\n ---------- RIDGE / L2 REGULARIZATION ----------")
print("Training Accuracy:", ridge_train_accuracy)
print("Testing Accuracy:", ridge_test_accuracy)

print("\n Classification Report:")
print(classification_report(y_test, ridge_test_pred))

print("\n Confusion Matrix:")
print(confusion_matrix(y_test, ridge_test_pred))
```

```python
# lasso like logistic regression
# Create L1-regularized Logistic Regression pipeline
lasso_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(penalty="l1", solver="liblinear", C=1.0, max_iter=5000))
])
```

```python
# Train Lasso-like model
lasso_model.fit(X_train, y_train)

# Predict training data
lasso_train_pred = lasso_model.predict(X_train)

# Predict testing data
lasso_test_pred = lasso_model.predict(X_test)

# Calculate training accuracy
lasso_train_accuracy = accuracy_score(y_train, lasso_train_pred)

# Calculate testing accuracy
lasso_test_accuracy = accuracy_score(y_test, lasso_test_pred)
```

```python
# Display Lasso results
print("\n ---------- LASSO / L1 REGULARIZATION ----------")
print("Training Accuracy:", lasso_train_accuracy)
print("Testing Accuracy:", lasso_test_accuracy)

# Create comparison table
results = pd.DataFrame({
    "Model": ["Ridge-like (L2)", "Lasso-like(L1)"],
    "Training Accuracy": [ridge_train_accuracy, lasso_train_accuracy],
    "Testing Accuracy": [ridge_test_accuracy, lasso_test_accuracy]
})
```

```python
# Display comparison table
print("\n --------- MODEL COMPARSION ----------")
print(results)
```

```python
# CHECK LASSO FEATURE SELECTION

# Extract Lasso coefficients
lasso_coefficients = lasso_model.named_steps["classifier"].coef_[0]

# Create coefficient table
coefficient_table = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": lasso_coefficients
})
```

```python
# Identify coefficients equal to zero
eliminated_features = coefficient_table[coefficient_table["Coefficient"] == 0]
retained_features = coefficient_table[coefficient_table["Coefficient"] != 0]

print("\n ---------- LASSO FEATURE SELECTION ----------")
print(f"Total Features: {len(coefficient_table)}")
print(f"Retained Features: {len(retained_features)}")
print(f"Eliminated Features (Coefficient = 0): {len(eliminated_features)}")
print("\nEliminated features:")
print(eliminated_features["Feature"].tolist())
```
